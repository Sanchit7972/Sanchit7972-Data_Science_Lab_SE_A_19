```python
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    elif n == 0:
        return 1
    else:
        fact = 1
        for i in range(1, n + 1):
            fact *= i
        return fact

# Example usage:
number = 5
print(f"The factorial of {number} is {factorial(number)}")

number = 0
print(f"The factorial of {number} is {factorial(number)}")

umber = -3
print(f"The factorial of {number} is {factorial(number)}")
```

# Task
Develop a basic expert system in Python that takes user input, applies a set of predefined rules, and displays a decision.

## Define Rules

### Subtask:
Define a set of simple rules for the expert system. For example, rules could be based on weather conditions to suggest an activity.


**Reasoning**:
The user wants to define a set of simple rules for an expert system. I will create a Python list of dictionaries, where each dictionary represents a rule with 'weather', 'temperature', and 'suggestion' keys, based on the examples provided.



```python
rules = [
    {
        "weather": "sunny",
        "temperature": "warm",
        "suggestion": "Go for a picnic"
    },
    {
        "weather": "rainy",
        "temperature": "cold",
        "suggestion": "Stay indoors and read a book"
    },
    {
        "weather": "sunny",
        "temperature": "cold",
        "suggestion": "Go for a brisk walk"
    },
    {
        "weather": "rainy",
        "temperature": "warm",
        "suggestion": "Visit a museum"
    },
    {
        "weather": "cloudy",
        "temperature": "mild",
        "suggestion": "Go to a cafe"
    },
    {
        "weather": "snowy",
        "temperature": "cold",
        "suggestion": "Build a snowman"
    }
]

print("Rules defined successfully:")
for rule in rules:
    print(rule)
```

## Implement Rule Engine

### Subtask:
Write Python code to implement a basic rule engine that takes input and applies the defined rules to make a decision. This can be done using if-elif-else statements or a dictionary-based approach.


**Reasoning**:
Implement the `make_decision` function as a rule engine, iterating through the predefined rules to find a match based on user-provided weather and temperature, and return a suggestion or a default message if no match is found. Then, test the function with sample inputs.



```python
def make_decision(current_weather, current_temperature):
    """
    Applies predefined rules to suggest an activity based on weather and temperature.
    """
    for rule in rules:
        if rule["weather"] == current_weather and rule["temperature"] == current_temperature:
            return rule["suggestion"]
    return "No specific suggestion based on current conditions."

# Test cases
print(f"Decision for sunny, warm: {make_decision('sunny', 'warm')}")
print(f"Decision for rainy, cold: {make_decision('rainy', 'cold')}")
print(f"Decision for cloudy, mild: {make_decision('cloudy', 'mild')}")
print(f"Decision for snowy, warm: {make_decision('snowy', 'warm')}") # Should return default
print(f"Decision for sunny, cold: {make_decision('sunny', 'cold')}")
```

## Get User Input

### Subtask:
Develop code to prompt the user for input relevant to the decision-making process (e.g., current weather, temperature).


## Get User Input

**Reasoning**:
To make the expert system interactive, I will prompt the user for the current weather and temperature. These inputs will then be used by the rule engine to determine a suggestion.

**Reasoning**:
Now I will write Python code to prompt the user for the current weather condition and temperature, storing these inputs in respective variables for subsequent use by the rule engine.



```python
user_weather = input("Enter the current weather (e.g., sunny, rainy, cloudy, snowy): ").lower()
user_temperature = input("Enter the current temperature (e.g., warm, cold, mild): ").lower()

print(f"You entered: Weather - {user_weather}, Temperature - {user_temperature}")
```

## Display Decision

### Subtask:
Integrate the rule engine with the input gathering step to display the decision made by the expert system to the user.


**Reasoning**:
I will integrate the previously defined `make_decision` function with the user's input (`user_weather` and `user_temperature`) to generate a suggestion and then display it to the user.



```python
final_suggestion = make_decision(user_weather, user_temperature)
print(f"\nExpert System Suggestion: {final_suggestion}")
```
