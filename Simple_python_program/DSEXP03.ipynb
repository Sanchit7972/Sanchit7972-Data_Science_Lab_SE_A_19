```python
import numpy as np
# Create a 3x3 square matrix
A = np.array([[1, 2, 3],
              [0, 1, 4],
              [5, 6, 0]])

# Compute determinant
det_A = np.linalg.det(A)

print("Matrix A:\n", A)
print("\nDeterminant of A:", det_A)
```

```python
# Find inverse of the matrix
A_inv = np.linalg.inv(A)

print("Inverse of Matrix A:\n", A_inv)
```

```python
# Coefficient matrix
coefficients = np.array([[2, 3],
                          [4, 5]])

# Constant terms
constants = np.array([10, 20])

# Solve the system of linear equations
solution = np.linalg.solve(coefficients, constants)

print("Solution:")
print("x =", solution[0])
print("y =", solution[1])
```
