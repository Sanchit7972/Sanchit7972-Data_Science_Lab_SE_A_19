```python
import numpy as np
import matplotlib.pyplot as plt

# Generate a normal distribution
mu, sigma = 0, 0.1 # mean and standard deviation
s = np.random.normal(mu, sigma, 1000)

# Plot the histogram
count, bins, ignored = plt.hist(s, 30, density=True)
plt.plot(bins, 1/(sigma * np.sqrt(2 * np.pi)) * np.exp( - (bins - mu)**2 / (2 * sigma**2) ), linewidth=2, color='r')
plt.title('Histogram of Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Assuming mu and sigma are already defined in the kernel from previous execution
# If not, they would need to be re-defined:
# mu, sigma = 0, 0.1

# Generate x-values for the plot
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 100)

# Calculate the corresponding y-values (Probability Density Function)
pdf = 1/(sigma * np.sqrt(2 * np.pi)) * np.exp( - (x - mu)**2 / (2 * sigma**2) )

# Plot the line graph
plt.plot(x, pdf, linewidth=2, color='r')
plt.title('Normal Distribution Probability Density Function')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.grid(True)
plt.show()
```

```python
import numpy as np

# Generate dataset for even numbers between 1 and 100
s_even = np.arange(2, 101, 2)

# Calculate statistics for the even numbers dataset
mean_s_even = np.mean(s_even)
median_s_even = np.median(s_even)
variance_s_even = np.var(s_even)
std_dev_s_even = np.std(s_even)

# Print the results
print(f"Statistics for even numbers between 1 and 100:")
print(f"Mean: {mean_s_even:.4f}")
print(f"Median: {median_s_even:.4f}")
print(f"Variance: {variance_s_even:.4f}")
print(f"Standard Deviation: {std_dev_s_even:.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Assuming s_even, mean_s_even, std_dev_s_even are already defined in the kernel

# Plot the histogram
count, bins, ignored = plt.hist(s_even, bins=10, density=True, alpha=0.6, color='g', label='Histogram of Even Numbers')

# Plot the mean as a vertical line
plt.axvline(mean_s_even, color='b', linestyle='dashed', linewidth=2, label=f'Mean: {mean_s_even:.2f}')

# Generate x-values for the theoretical normal distribution curve
x_normal = np.linspace(min(s_even) - 10, max(s_even) + 10, 1000)

# Calculate the corresponding y-values (Probability Density Function)
# using the mean and standard deviation of the s_even dataset
pdf_normal = 1/(std_dev_s_even * np.sqrt(2 * np.pi)) * np.exp( - (x_normal - mean_s_even)**2 / (2 * std_dev_s_even**2) )

# Plot the theoretical normal distribution curve
plt.plot(x_normal, pdf_normal, linewidth=2, color='r', label='Normal Distribution Curve')

plt.title('Histogram of Even Numbers with Mean and Normal Distribution Curve')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np

# Number of coin tosses to simulate
num_tosses = 10000

# Simulate coin tosses (0 for Tails, 1 for Heads)
toss_results = np.random.randint(0, 2, num_tosses)

# Count occurrences of Heads and Tails
heads_count = np.sum(toss_results == 1)
tails_count = np.sum(toss_results == 0)

# Calculate probabilities
probability_heads = heads_count / num_tosses
probability_tails = tails_count / num_tosses

print(f"Total Coin Tosses: {num_tosses}")
print(f"Heads Count: {heads_count}")
print(f"Tails Count: {tails_count}")
print(f"Probability of Heads: {probability_heads:.4f}")
print(f"Probability of Tails: {probability_tails:.4f}")
```
