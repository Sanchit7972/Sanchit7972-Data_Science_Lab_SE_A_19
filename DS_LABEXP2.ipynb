```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    print("BFS Traversal:")
    while queue:
        node = queue.popleft()
        print(node, end=" ")

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

# Example graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

bfs(graph, 'A')
```

```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    visited.add(start)
    print(start, end=" ")

    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)

print("\nDFS Traversal:")
dfs(graph, 'A')
```

```python
import heapq

def a_star(graph, heuristic, start, goal):
    # Priority queue: (f_cost, g_cost, current_node, path)
    open_list = []
    heapq.heappush(open_list, (heuristic[start], 0, start, [start]))

    closed_set = set()

    while open_list:
        f_cost, g_cost, current, path = heapq.heappop(open_list)

        # Goal check
        if current == goal:
            return path, g_cost

        if current in closed_set:
            continue

        closed_set.add(current)

        # Explore neighbors
        for neighbor, cost in graph[current]:
            if neighbor not in closed_set:
                new_g = g_cost + cost
                new_f = new_g + heuristic[neighbor]
                heapq.heappush(
                    open_list,
                    (new_f, new_g, neighbor, path + [neighbor])
                )

    return None, float('inf')

graph = {
    'A': [('B', 1), ('C', 3)],
    'B': [('D', 3), ('E', 1)],
    'C': [('F', 5)],
    'D': [],
    'E': [('F', 2)],
    'F': []
}

heuristic = {
    'A': 6,
    'B': 4,
    'C': 4,
    'D': 3,
    'E': 2,
    'F': 0
}
path, cost = a_star(graph, heuristic, 'A', 'F')

print("A* Path:", path)
print("Total Cost:", cost)
```

```python
def minimax(depth, node_index, is_maximizing, values):
    # Terminal condition: leaf node
    if depth == 0:
        return values[node_index]

    if is_maximizing:
        best_value = float('-inf')
        for i in range(2):
            value = minimax(depth - 1, node_index * 2 + i, False, values)
            best_value = max(best_value, value)
        return best_value
    else:
        best_value = float('inf')
        for i in range(2):
            value = minimax(depth - 1, node_index * 2 + i, True, values)
            best_value = min(best_value, value)
        return best_value
# Leaf node values (utilities at depth = 0)
values = [3, 5, 6, 9, 1, 2, 0, -1]

# Depth of the game tree
depth = 3
optimal_value = minimax(depth, 0, True, values)
print("Optimal value using Minimax:", optimal_value)
```
